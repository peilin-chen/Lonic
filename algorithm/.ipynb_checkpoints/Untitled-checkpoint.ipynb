```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Absolute latency data (us)
# -----------------------------
datasets = ["DVS128-Gesture", "DVS-CIFAR10", "CIFAR-10", "CIFAR-100"]
methods = ["OTTT(sWS)", "SLTT(sWS)", "SpikON(LTTT+sWCTT)", "TESS(sWS)"]

# Element-wise computation (us)
elementwise = np.array([
	[476608.3645, 363056.4815, 246888.9912, 439674.3790],
	[158641.3509, 127292.9435,  70749.08511, 147047.5352],
	[ 92878.41177, 74785.9840,   40608.00694, 86576.88811],
	[ 93444.79832, 74789.2080,   40624.59722, 86436.29371],
])

# Matrix multiplication (us)
matmul = np.array([
	[122981, 93817, 93483, 129859],
	[25216, 20152, 20290, 26197],
	[12670, 10333, 10278, 13441],
	[12662, 10353, 10314, 13451],
])

# -----------------------------
# Plot: grouped stacked bars
# -----------------------------
plt.figure(figsize=(14, 4.5))

bar_w = 0.75
inner_gap = 0.15
group_gap = 1.6

x_positions = []
x_labels = []
group_centers = []

x = 0.0
for di in range(len(datasets)):
	start_x = x
	for mi in range(len(methods)):
		x_positions.append(x)
		x_labels.append(methods[mi])
		x += bar_w + inner_gap
	end_x = x - (bar_w + inner_gap)
	group_centers.append((start_x + end_x) / 2.0)
	x += group_gap

x_positions = np.array(x_positions)

ew_flat = elementwise.reshape(-1)
mm_flat = matmul.reshape(-1)

# Bottom: Matrix multiplication
plt.bar(
	x_positions, mm_flat, width=bar_w,
	edgecolor="black", linewidth=1.0,
	hatch="xx", label="Matrix multiplication"
)

# Top: Element-wise computation
plt.bar(
	x_positions, ew_flat, width=bar_w, bottom=mm_flat,
	edgecolor="black", linewidth=1.0,
	hatch="++", label="Element-wise computation"
)

# Group separators (dashed vertical lines)
for gi in range(1, len(datasets)):
	sep_x = (group_centers[gi-1] + group_centers[gi]) / 2.0
	plt.axvline(sep_x, linestyle="--", linewidth=2.0, color="gray")

# Dataset titles above each group
y_top = (ew_flat + mm_flat).max() * 1.05
for c, d in zip(group_centers, datasets):
	plt.text(c, y_top, d, ha="center", va="bottom",
			 fontsize=12, fontweight="bold")

plt.ylabel("Latency (µs)", fontsize=12)
plt.xticks(x_positions, x_labels, rotation=25, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.legend(loc="upper right")

plt.tight_layout()
plt.show()
```
