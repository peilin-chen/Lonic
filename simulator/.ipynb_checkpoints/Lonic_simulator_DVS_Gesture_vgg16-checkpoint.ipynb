```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math
from typing import Tuple
from typing import Callable
from dataclasses import dataclass

def estimate_Conv2d_layer_1(class_num, in_ch, out_ch, kernel_size, x_shape, stride=1, padding=1, dilation=1, groups=1, bias=False, simd_parallel=32, on_chip_quantizer_capacity=64, data_compressor_capacity=4*64, 
                              PE_array_type0_capacity=64, on_chip_dequantizer_capacity=64, pre_trace_calculator_capacity=64, post_trace_calculator_capacity=64, on_chip_mac_capacity=64,
                             PE_array_type1_capacity=16, PE_array_type2_capacity=64, PE_array_type0_pair_sparsity=0.0):
    N, Cin, Hin, Win = x_shape
    Cout = out_ch
    Kh, Kw = (kernel_size, kernel_size)
    sh, sw = (stride, stride)
    ph, pw = (padding, padding)
    dh, dw = (dilation, dilation)

    Hout = (Hin + 2*ph - dh*(Kh-1) - 1)//sh + 1
    Wout = (Win + 2*pw - dw*(Kw-1) - 1)//sw + 1
 
    K = (Cin // groups) * Kh * Kw
    M = N * Cout * Hout * Wout

    inst = {}
    inst["add"] = 0
    inst["sub"] = 0 
    inst["mul"] = 0 
    inst["div"] = 0 
    inst["sqrt"] = 0 
    inst["min"] = 0 
    inst["max"] = 0 
    inst["lif"] = 0 
    inst["reset"] = 0 
    inst["sg"] = 0 
    inst["load"] = 0 
    inst["store"] = 0 
    
    # forward
    # 1: weight centralization
    # compute sum
    inst["load"] += math.ceil(Cout*K/simd_parallel)
    inst["add"] += math.ceil(Cout*K/simd_parallel)
    
    # compute mean
    inst["load"] += math.ceil(1/simd_parallel)
    inst["div"] += math.ceil(Cout/simd_parallel)
    
    # compute weight centralization
    inst["load"] += math.ceil(Cout*K/simd_parallel)
    inst["sub"] += math.ceil(Cout*K/simd_parallel)
    
    # compute scaled weight
    inst["load"] += math.ceil(1/simd_parallel)
    inst["mul"] += math.ceil(Cout*K/simd_parallel)
    inst["store"] += math.ceil(Cout*K/simd_parallel)

    # on-chip quantizer
    inst["load"] += math.ceil(Cout*K/simd_parallel)
    inst["max"] += math.ceil(Cout*K/simd_parallel)
    inst["store"] += math.ceil(Cout/simd_parallel)

    on_chip_quantizer_computation = math.ceil(Cout*K/on_chip_quantizer_capacity)

    # data compressor
    data_compressor_computation = 0
    
    on_chip_mac_computation = math.ceil(K * M / on_chip_mac_capacity / 2) # reuse pre-trace calculator
    
    pe_array_type0_computation = 0

    # on-chip dequantizer
    on_chip_dequantizer_computation = 0

    # pre-trace calculator
    pre_trace_calculator_computation = math.ceil(N*Cin*Hin*Win / pre_trace_calculator_capacity)

    # post-trace calculator
    post_trace_calculator_computation = math.ceil(M / post_trace_calculator_capacity)

    # Mult. & Add ( mem = leak_factor * mem + output)
    on_chip_mac_computation = math.ceil(M / on_chip_mac_capacity / 2)

    # lif
    inst["load"] += math.ceil(M / simd_parallel)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["lif"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    # reset
    inst["load"] += math.ceil(M / simd_parallel)
    inst["load"] += math.ceil(M / simd_parallel)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["reset"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    # on-chip quantizer for B matrix
    on_chip_quantizer_computation += 0

    on_chip_mac_computation += math.ceil(N*class_num*Cout*Hout*Wout / on_chip_mac_capacity / 2)
    
    pe_array_type1_computation = 0

    # on-chip dequantizer
    on_chip_dequantizer_computation += 0

    # cross entropy
    # m = max_j z_j
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["max"] += math.ceil(N * class_num / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    # r_j = z_j - m
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["sub"] += math.ceil(N * class_num / simd_parallel)
    inst["store"] += math.ceil(N * class_num / simd_parallel)
    # e_j = exp(r_j)
    # exp(r) ≈ 1 + r*(1 + r*(1/2 + r*(1/6 + r*(1/24 + r*(1/120)))))
    for _ in range(5):
        inst["load"] += math.ceil(N * class_num / simd_parallel)
        inst["load"] += math.ceil(1 / simd_parallel)
        inst["load"] += math.ceil(1 / simd_parallel)
        inst["mul"] += math.ceil(N * class_num / simd_parallel)
        inst["add"] += math.ceil(N * class_num / simd_parallel)
        inst["store"] += math.ceil(N * class_num / simd_parallel)
    # sum_e = ∑ e_j
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["add"]  += math.ceil(N * class_num / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    # lse = log(sum_e) + m 
    # y = (s-1)/(s+1)
    # ln s ≈ 2*(y + y^3/3 + y^5/5 + y^7/7)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["sub"] += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["add"] += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["div"] += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^2
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^3
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^4
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^5
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^6
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^7
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(4*N / simd_parallel)
    inst["add"] += math.ceil(4*N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["add"]  += math.ceil(N / simd_parallel) # + m
    inst["store"] += math.ceil(N / simd_parallel)

    # loss_i = lse - z_y
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["sub"]  += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)

    # backward
    # m = max_j z_j
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["max"]  += math.ceil(N * class_num / simd_parallel)
    inst["store"]+= math.ceil(N / simd_parallel)

    # r = z - m
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["sub"]  += math.ceil(N * class_num / simd_parallel)
    inst["store"]+= math.ceil(N * class_num / simd_parallel)

    # e = exp(r)
    for _ in range(5):
        inst["load"] += math.ceil(N * class_num / simd_parallel)
        inst["load"] += math.ceil(1 / simd_parallel)
        inst["load"] += math.ceil(1 / simd_parallel)
        inst["mul"] += math.ceil(N * class_num / simd_parallel)
        inst["add"] += math.ceil(N * class_num / simd_parallel)
        inst["store"] += math.ceil(N * class_num / simd_parallel)

    # sum_e = ∑_j e_j
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["add"]  += math.ceil(N * class_num / simd_parallel)
    inst["store"]+= math.ceil(N / simd_parallel)

    # p = softmax = e / sum_e
    inst["load"] += math.ceil(N * class_num / simd_parallel)   # load e
    inst["load"] += math.ceil(N / simd_parallel)   # load sum_e
    inst["div"]  += math.ceil(N * class_num / simd_parallel)   # e / sum_e
    inst["store"]+= math.ceil(N * class_num / simd_parallel)   # store p

    # grad = p - one_hot(y) 
    inst["load"] += math.ceil(N * class_num / simd_parallel) 
    inst["load"] += math.ceil(N * class_num / simd_parallel)  
    inst["sub"]  += math.ceil(N * class_num / simd_parallel)  
    inst["store"]+= math.ceil(N * class_num / simd_parallel) 

    # on-chip quantizer 
    on_chip_quantizer_computation += 0

    on_chip_mac_computation += math.ceil(N*Cout*Hout*Wout*class_num / on_chip_mac_capacity / 2)
    
    pe_array_type2_computation = 0

    # on-chip dequantizer (m)
    on_chip_dequantizer_computation += 0

    # calculate delta W_pre
    inst["load"] += math.ceil(M / simd_parallel)
    inst["sg"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)
    
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["load"] += math.ceil(M / simd_parallel)
    inst["mul"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    inst["load"] += math.ceil(M / simd_parallel)
    inst["load"] += math.ceil(M / simd_parallel)
    inst["mul"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    on_chip_mac_computation += math.ceil(Cout*Cin*Kh*Kw*Hout*Wout / on_chip_mac_capacity / 2)

    # calculate delta W_post 
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["load"] += math.ceil(M / simd_parallel)
    inst["mul"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    inst["load"] += math.ceil(M / simd_parallel)
    inst["load"] += math.ceil(M / simd_parallel)
    inst["mul"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    on_chip_mac_computation += math.ceil(Cout*Cin*Kh*Kw*Hout*Wout / on_chip_mac_capacity / 2)

    # calculate delta W
    inst["load"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["load"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["add"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["store"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)

    # Wt = γ · (w - mean(w))
    # dW = γ · (dWt - mean(dWt))  dγ = sum(dWt · (w - mean(w)))
    # mean(dWt)
    inst["load"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["add"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["div"] += math.ceil(Cout / simd_parallel)

    # dW = γ · (dWt - mean(dWt))
    inst["sub"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)   # G - mean(G)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["mul"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)   # * gamma
    inst["store"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)

    # dγ = sum(dWt · (w - mean(w)))
    # compute sum(w)
    inst["load"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["add"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    # compute mean(w)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["div"] += math.ceil(Cout / simd_parallel)
    # compute w - mean(w)
    inst["sub"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    # dγ = sum(dWt · (w - mean(w)))
    inst["mul"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["add"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["store"] += math.ceil(1 / simd_parallel)    
    
    total_instructions = sum(inst.values())
    
    return total_instructions, on_chip_quantizer_computation, data_compressor_computation, pe_array_type0_computation, pe_array_type1_computation, pe_array_type2_computation, on_chip_dequantizer_computation, pre_trace_calculator_computation, post_trace_calculator_computation, on_chip_mac_computation 

def estimate_Conv2d_layer_2_8(class_num, in_ch, out_ch, kernel_size, x_shape, stride=1, padding=1, dilation=1, groups=1, bias=False, simd_parallel=32, on_chip_quantizer_capacity=64, data_compressor_capacity=4*64, 
                              PE_array_type0_capacity=64, on_chip_dequantizer_capacity=64, pre_trace_calculator_capacity=64, post_trace_calculator_capacity=64, on_chip_mac_capacity=64,
                             PE_array_type1_capacity=16, PE_array_type2_capacity=64, PE_array_type0_pair_sparsity=0.90):
    N, Cin, Hin, Win = x_shape
    Cout = out_ch
    Kh, Kw = (kernel_size, kernel_size)
    sh, sw = (stride, stride)
    ph, pw = (padding, padding)
    dh, dw = (dilation, dilation)

    Hout = (Hin + 2*ph - dh*(Kh-1) - 1)//sh + 1
    Wout = (Win + 2*pw - dw*(Kw-1) - 1)//sw + 1

    K = (Cin // groups) * Kh * Kw
    M = N * Cout * Hout * Wout

    inst = {}
    inst["add"] = 0
    inst["sub"] = 0 
    inst["mul"] = 0 
    inst["div"] = 0 
    inst["sqrt"] = 0 
    inst["min"] = 0 
    inst["max"] = 0 
    inst["lif"] = 0 
    inst["reset"] = 0 
    inst["sg"] = 0 
    inst["load"] = 0 
    inst["store"] = 0 
    
    # forward
    # 1: weight centralization
    # compute sum
    inst["load"] += math.ceil(Cout*K/simd_parallel)
    inst["add"] += math.ceil(Cout*K/simd_parallel)
    
    # compute mean
    inst["load"] += math.ceil(1/simd_parallel)
    inst["div"] += math.ceil(Cout/simd_parallel)
    
    # compute weight centralization
    inst["load"] += math.ceil(Cout*K/simd_parallel)
    inst["sub"] += math.ceil(Cout*K/simd_parallel)
    
    # compute scaled weight
    inst["load"] += math.ceil(1/simd_parallel)
    inst["mul"] += math.ceil(Cout*K/simd_parallel)
    inst["store"] += math.ceil(Cout*K/simd_parallel)

    # on-chip quantizer
    inst["load"] += math.ceil(Cout*K/simd_parallel)
    inst["max"] += math.ceil(Cout*K/simd_parallel)
    inst["store"] += math.ceil(Cout/simd_parallel)

    on_chip_quantizer_computation = math.ceil(Cout*K/on_chip_quantizer_capacity)

    # data compressor
    data_compressor_computation = math.ceil(M*K/data_compressor_capacity)
    
    # PE array type0
    batches = math.ceil(M / PE_array_type0_capacity / 4) # if weight bitwidth=4 -> use /4, =8 -> use /2, =16 -> use /1
    pe_array_type0_computation = math.ceil(batches * K * (1-PE_array_type0_pair_sparsity))

    # on-chip dequantizer
    on_chip_dequantizer_computation = math.ceil(M / on_chip_dequantizer_capacity)

    # pre-trace calculator
    pre_trace_calculator_computation = math.ceil(N*Cin*Hin*Win / pre_trace_calculator_capacity)

    # post-trace calculator
    post_trace_calculator_computation = math.ceil(M / post_trace_calculator_capacity)

    # Mult. & Add ( mem = leak_factor * mem + output)
    on_chip_mac_computation = math.ceil(M / on_chip_mac_capacity / 2)

    # lif
    inst["load"] += math.ceil(M / simd_parallel)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["lif"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    # reset
    inst["load"] += math.ceil(M / simd_parallel)
    inst["load"] += math.ceil(M / simd_parallel)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["reset"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    # on-chip quantizer for B matrix
    on_chip_quantizer_computation += math.ceil(class_num*Cout*Hout*Wout / on_chip_quantizer_capacity)

    # PE array type1
    batches = math.ceil(N * class_num / PE_array_type1_capacity / 8)
    pe_array_type1_computation = batches * Cout * Hout * Wout

    # on-chip dequantizer
    on_chip_dequantizer_computation += math.ceil(N * class_num / on_chip_dequantizer_capacity)

    # cross entropy
    # m = max_j z_j
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["max"] += math.ceil(N * class_num / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    # r_j = z_j - m
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["sub"] += math.ceil(N * class_num / simd_parallel)
    inst["store"] += math.ceil(N * class_num / simd_parallel)
    # e_j = exp(r_j)
    # exp(r) ≈ 1 + r*(1 + r*(1/2 + r*(1/6 + r*(1/24 + r*(1/120)))))
    for _ in range(5):
        inst["load"] += math.ceil(N * class_num / simd_parallel)
        inst["load"] += math.ceil(1 / simd_parallel)
        inst["load"] += math.ceil(1 / simd_parallel)
        inst["mul"] += math.ceil(N * class_num / simd_parallel)
        inst["add"] += math.ceil(N * class_num / simd_parallel)
        inst["store"] += math.ceil(N * class_num / simd_parallel)
    # sum_e = ∑ e_j
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["add"]  += math.ceil(N * class_num / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    # lse = log(sum_e) + m 
    # y = (s-1)/(s+1)
    # ln s ≈ 2*(y + y^3/3 + y^5/5 + y^7/7)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["sub"] += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["add"] += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["div"] += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^2
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^3
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^4
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^5
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^6
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^7
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(4*N / simd_parallel)
    inst["add"] += math.ceil(4*N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["add"]  += math.ceil(N / simd_parallel) # + m
    inst["store"] += math.ceil(N / simd_parallel)

    # loss_i = lse - z_y
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["sub"]  += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)

    # backward
    # m = max_j z_j
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["max"]  += math.ceil(N * class_num / simd_parallel)
    inst["store"]+= math.ceil(N / simd_parallel)

    # r = z - m
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["sub"]  += math.ceil(N * class_num / simd_parallel)
    inst["store"]+= math.ceil(N * class_num / simd_parallel)

    # e = exp(r)
    for _ in range(5):
        inst["load"] += math.ceil(N * class_num / simd_parallel)
        inst["load"] += math.ceil(1 / simd_parallel)
        inst["load"] += math.ceil(1 / simd_parallel)
        inst["mul"] += math.ceil(N * class_num / simd_parallel)
        inst["add"] += math.ceil(N * class_num / simd_parallel)
        inst["store"] += math.ceil(N * class_num / simd_parallel)

    # sum_e = ∑_j e_j
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["add"]  += math.ceil(N * class_num / simd_parallel)
    inst["store"]+= math.ceil(N / simd_parallel)

    # p = softmax = e / sum_e
    inst["load"] += math.ceil(N * class_num / simd_parallel)   # load e
    inst["load"] += math.ceil(N / simd_parallel)   # load sum_e
    inst["div"]  += math.ceil(N * class_num / simd_parallel)   # e / sum_e
    inst["store"]+= math.ceil(N * class_num / simd_parallel)   # store p

    # grad = p - one_hot(y) 
    inst["load"] += math.ceil(N * class_num / simd_parallel) 
    inst["load"] += math.ceil(N * class_num / simd_parallel)  
    inst["sub"]  += math.ceil(N * class_num / simd_parallel)  
    inst["store"]+= math.ceil(N * class_num / simd_parallel) 

    # on-chip quantizer 
    on_chip_quantizer_computation += math.ceil(N * class_num / on_chip_quantizer_capacity)

    # PE array type2
    batches = math.ceil(M / PE_array_type2_capacity / 2)
    pe_array_type2_computation = batches * class_num

    # on-chip dequantizer (m)
    on_chip_dequantizer_computation += math.ceil(M / on_chip_dequantizer_capacity)

    # calculate delta W_pre
    inst["load"] += math.ceil(M / simd_parallel)
    inst["sg"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)
    
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["load"] += math.ceil(M / simd_parallel)
    inst["mul"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    inst["load"] += math.ceil(M / simd_parallel)
    inst["load"] += math.ceil(M / simd_parallel)
    inst["mul"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    on_chip_mac_computation += math.ceil(Cout*Cin*Kh*Kw*Hout*Wout / on_chip_mac_capacity / 2)

    # calculate delta W_post 
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["load"] += math.ceil(M / simd_parallel)
    inst["mul"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    inst["load"] += math.ceil(M / simd_parallel)
    inst["load"] += math.ceil(M / simd_parallel)
    inst["mul"] += math.ceil(M / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    on_chip_mac_computation += math.ceil(Cout*Cin*Kh*Kw*Hout*Wout / on_chip_mac_capacity / 2)

    # calculate delta W
    inst["load"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["load"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["add"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["store"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)

    # Wt = γ · (w - mean(w))
    # dW = γ · (dWt - mean(dWt))  dγ = sum(dWt · (w - mean(w)))
    # mean(dWt)
    inst["load"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["add"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["div"] += math.ceil(Cout / simd_parallel)

    # dW = γ · (dWt - mean(dWt))
    inst["sub"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)   # G - mean(G)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["mul"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)   # * gamma
    inst["store"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)

    # dγ = sum(dWt · (w - mean(w)))
    # compute sum(w)
    inst["load"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["add"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    # compute mean(w)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["div"] += math.ceil(Cout / simd_parallel)
    # compute w - mean(w)
    inst["sub"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    # dγ = sum(dWt · (w - mean(w)))
    inst["mul"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["add"] += math.ceil(Cin*Cout*Kh*Kw / simd_parallel)
    inst["store"] += math.ceil(1 / simd_parallel)    
    
    total_instructions = sum(inst.values())
    
    return total_instructions, on_chip_quantizer_computation, data_compressor_computation, pe_array_type0_computation, pe_array_type1_computation, pe_array_type2_computation, on_chip_dequantizer_computation, pre_trace_calculator_computation, post_trace_calculator_computation, on_chip_mac_computation  

def estimate_Linear_layer(class_num, in_ch, out_ch, x_shape, bias=True, simd_parallel=32, on_chip_quantizer_capacity=64, data_compressor_capacity=4*64, 
                              PE_array_type0_capacity=64, on_chip_dequantizer_capacity=64, pre_trace_calculator_capacity=64, post_trace_calculator_capacity=64, on_chip_mac_capacity=64,
                             PE_array_type1_capacity=16, PE_array_type2_capacity=64, PE_array_type0_pair_sparsity=0.0):
    N, Cin = x_shape
    Cout = out_ch

    K = Cin
    M = N * Cout

    inst = {}
    inst["add"] = 0
    inst["sub"] = 0 
    inst["mul"] = 0 
    inst["div"] = 0 
    inst["sqrt"] = 0 
    inst["min"] = 0 
    inst["max"] = 0 
    inst["lif"] = 0 
    inst["reset"] = 0 
    inst["sg"] = 0 
    inst["load"] = 0 
    inst["store"] = 0 
    
    # forward
    # add bias
    inst["load"]  += math.ceil(M / simd_parallel) if bias is not None else 0
    inst["load"] += math.ceil(Cout / simd_parallel) if bias is not None else 0
    inst["add"]   += math.ceil(M / simd_parallel) if bias is not None else 0
    inst["store"] += math.ceil(M / simd_parallel) if bias is not None else 0

    # on-chip quantizer
    inst["load"] += math.ceil(Cout*K/simd_parallel)
    inst["max"] += math.ceil(Cout*K/simd_parallel)
    inst["store"] += math.ceil(Cout/simd_parallel)

    on_chip_quantizer_computation = math.ceil(Cout*K/on_chip_quantizer_capacity)

    # data compressor
    data_compressor_computation = 0
    
    on_chip_mac_computation = math.ceil(N*Cout*Cin / on_chip_mac_capacity / 2)
    pe_array_type0_computation = 0

    # on-chip dequantizer
    on_chip_dequantizer_computation = 0

    # pre-trace calculator
    pre_trace_calculator_computation = 0

    # post-trace calculator
    post_trace_calculator_computation = 0

    # Mult. & Add ( mem = leak_factor * mem + output)
    on_chip_mac_computation += 0

    # lif
    #inst["load"] += math.ceil(M / simd_parallel)
    #inst["load"] += math.ceil(1 / simd_parallel)
    #inst["lif"] += math.ceil(M / simd_parallel)
    #inst["store"] += math.ceil(M / simd_parallel)

    # reset
    #inst["load"] += math.ceil(M / simd_parallel)
    #inst["load"] += math.ceil(M / simd_parallel)
    #inst["load"] += math.ceil(1 / simd_parallel)
    #inst["reset"] += math.ceil(M / simd_parallel)
    #inst["store"] += math.ceil(M / simd_parallel)

    # on-chip quantizer for B matrix
    on_chip_quantizer_computation += 0

    # PE array type1
    pe_array_type1_computation = 0

    # on-chip dequantizer
    on_chip_dequantizer_computation += 0

    # cross entropy
    # m = max_j z_j
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["max"] += math.ceil(N * class_num / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    # r_j = z_j - m
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["sub"] += math.ceil(N * class_num / simd_parallel)
    inst["store"] += math.ceil(N * class_num / simd_parallel)
    # e_j = exp(r_j)
    # exp(r) ≈ 1 + r*(1 + r*(1/2 + r*(1/6 + r*(1/24 + r*(1/120)))))
    for _ in range(5):
        inst["load"] += math.ceil(N * class_num / simd_parallel)
        inst["load"] += math.ceil(1 / simd_parallel)
        inst["load"] += math.ceil(1 / simd_parallel)
        inst["mul"] += math.ceil(N * class_num / simd_parallel)
        inst["add"] += math.ceil(N * class_num / simd_parallel)
        inst["store"] += math.ceil(N * class_num / simd_parallel)
    # sum_e = ∑ e_j
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["add"]  += math.ceil(N * class_num / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    # lse = log(sum_e) + m 
    # y = (s-1)/(s+1)
    # ln s ≈ 2*(y + y^3/3 + y^5/5 + y^7/7)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["sub"] += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["add"] += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["div"] += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^2
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^3
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^4
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^5
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^6
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel) # y^7
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(4*N / simd_parallel)
    inst["add"] += math.ceil(4*N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(1 / simd_parallel)
    inst["mul"] += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)
    
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["add"]  += math.ceil(N / simd_parallel) # + m
    inst["store"] += math.ceil(N / simd_parallel)

    # loss_i = lse - z_y
    inst["load"] += math.ceil(N / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["sub"]  += math.ceil(N / simd_parallel)
    inst["store"] += math.ceil(N / simd_parallel)

    # backward
    # m = max_j z_j
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["max"]  += math.ceil(N * class_num / simd_parallel)
    inst["store"]+= math.ceil(N / simd_parallel)

    # r = z - m
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["load"] += math.ceil(N / simd_parallel)
    inst["sub"]  += math.ceil(N * class_num / simd_parallel)
    inst["store"]+= math.ceil(N * class_num / simd_parallel)

    # e = exp(r)
    for _ in range(5):
        inst["load"] += math.ceil(N * class_num / simd_parallel)
        inst["load"] += math.ceil(1 / simd_parallel)
        inst["load"] += math.ceil(1 / simd_parallel)
        inst["mul"] += math.ceil(N * class_num / simd_parallel)
        inst["add"] += math.ceil(N * class_num / simd_parallel)
        inst["store"] += math.ceil(N * class_num / simd_parallel)

    # sum_e = ∑_j e_j
    inst["load"] += math.ceil(N * class_num / simd_parallel)
    inst["add"]  += math.ceil(N * class_num / simd_parallel)
    inst["store"]+= math.ceil(N / simd_parallel)

    # p = softmax = e / sum_e
    inst["load"] += math.ceil(N * class_num / simd_parallel)   # load e
    inst["load"] += math.ceil(N / simd_parallel)   # load sum_e
    inst["div"]  += math.ceil(N * class_num / simd_parallel)   # e / sum_e
    inst["store"]+= math.ceil(N * class_num / simd_parallel)   # store p

    # grad = p - one_hot(y) 
    inst["load"] += math.ceil(N * class_num / simd_parallel) 
    inst["load"] += math.ceil(N * class_num / simd_parallel)  
    inst["sub"]  += math.ceil(N * class_num / simd_parallel)  
    inst["store"]+= math.ceil(N * class_num / simd_parallel) 

    # on-chip quantizer 
    on_chip_quantizer_computation += 0

    # PE array type2
    pe_array_type2_computation = 0

    # on-chip dequantizer (m)
    on_chip_dequantizer_computation += 0

    on_chip_mac_computation = math.ceil(Cout*Cin / on_chip_mac_capacity / 2)

    if bias is not None:
        inst["load"]  += math.ceil(Cout * N / simd_parallel)                            
        inst["add"]   += math.ceil(Cout * N / simd_parallel)          
        inst["store"] += math.ceil(Cout / simd_parallel)         
    
    total_instructions = sum(inst.values())
    
    return total_instructions, on_chip_quantizer_computation, data_compressor_computation, pe_array_type0_computation, pe_array_type1_computation, pe_array_type2_computation, on_chip_dequantizer_computation, pre_trace_calculator_computation, post_trace_calculator_computation, on_chip_mac_computation  

def estimate_MaxPool2d_layer(x_shape, kernel_size, stride, padding, simd_parallel=32):
    N, C, Hin, Win = x_shape

    Kh, Kw = (kernel_size, kernel_size)
    sh, sw = (stride, stride)
    ph, pw = (padding, padding)

    # output feature map
    Hout = (Hin + 2*ph - (Kh - 1) - 1) // sh + 1
    Wout = (Win + 2*pw - (Kw - 1) - 1) // sw + 1

    M = N * C * Hout * Wout
    K = Kh * Kw

    inst = {}
    inst["add"] = 0
    inst["sub"] = 0 
    inst["mul"] = 0 
    inst["div"] = 0 
    inst["sqrt"] = 0 
    inst["min"] = 0 
    inst["max"] = 0 
    inst["lif"] = 0 
    inst["reset"] = 0 
    inst["sg"] = 0 
    inst["load"] = 0 
    inst["store"] = 0

    # forward
    inst["load"] += math.ceil(M * K / simd_parallel)
    inst["max"] += math.ceil(M * K / simd_parallel)
    inst["store"] += math.ceil(M / simd_parallel)

    # no backward for maxpool layer (w/o learning param) since it is local learning
    
    total_instructions = sum(inst.values())

    return total_instructions
    
########################################### VGG Network define ###########################################

def estimate_vgg_dvs_gesture_training_one_iteration():
    total_instructions = 0
    on_chip_quantizer_computation = 0
    data_compressor_computation = 0
    pe_array_type0_computation = 0
    pe_array_type1_computation = 0
    pe_array_type2_computation = 0
    on_chip_dequantizer_computation = 0
    pre_trace_calculator_computation = 0
    post_trace_calculator_computation = 0
    on_chip_mac_computation = 0

    param_size = 0

    # input: (1, 2, 128, 128)
    # conv1: (2, 64, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 2, 128, 128) output->(1, 64, 128, 128)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_1(class_num=11, in_ch=2, out_ch=64, kernel_size=3, x_shape=(1, 2, 128, 128), PE_array_type0_pair_sparsity=0.0)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 2*64*3*3

    # conv2: (64, 64, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 64, 128, 128) output->(1, 64, 128, 128)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_2_8(class_num=11, in_ch=64, out_ch=64, kernel_size=3, x_shape=(1, 64, 128, 128), PE_array_type0_pair_sparsity=0.897254)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 64*64*3*3

    # maxpool1: (kernel_size=2, stride=2) & input->(1, 64, 128, 128) output->(1, 64, 64, 64)
    x0 = estimate_MaxPool2d_layer(x_shape=(1, 64, 128, 128), kernel_size=2, stride=2, padding=0)
    total_instructions += x0

    # conv3: (64, 128, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 64, 64, 64) output->(1, 128, 64, 64)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_2_8(class_num=11, in_ch=64, out_ch=128, kernel_size=3, x_shape=(1, 64, 64, 64), PE_array_type0_pair_sparsity=0.897254)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 64*128*3*3

    # conv4: (128, 128, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 128, 64, 64) output->(1, 128, 64, 64)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_2_8(class_num=11, in_ch=128, out_ch=128, kernel_size=3, x_shape=(1, 128, 64, 64), PE_array_type0_pair_sparsity=0.897254)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 128*128*3*3

    # maxpool2: (kernel_size=2, stride=2) & input->(1, 128, 64, 64) output->(1, 128, 32, 32)
    x0 = estimate_MaxPool2d_layer(x_shape=(1, 128, 64, 64), kernel_size=2, stride=2, padding=0)
    total_instructions += x0

    # conv5: (128, 256, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 128, 32, 32) output->(1, 256, 32, 32)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_2_8(class_num=11, in_ch=128, out_ch=256, kernel_size=3, x_shape=(1, 128, 32, 32), PE_array_type0_pair_sparsity=0.897254)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 128*256*3*3

    # conv6: (256, 256, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 256, 32, 32) output->(1, 256, 32, 32)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_2_8(class_num=11, in_ch=256, out_ch=256, kernel_size=3, x_shape=(1, 256, 32, 32), PE_array_type0_pair_sparsity=0.897254)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 256*256*3*3

    # conv7: (256, 256, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 256, 32, 32) output->(1, 256, 32, 32)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_2_8(class_num=11, in_ch=256, out_ch=256, kernel_size=3, x_shape=(1, 256, 32, 32), PE_array_type0_pair_sparsity=0.897254)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 256*256*3*3

    # maxpool3: (kernel_size=2, stride=2) & input->(1, 256, 32, 32) output->(1, 256, 16, 16)
    x0 = estimate_MaxPool2d_layer(x_shape=(1, 256, 32, 32), kernel_size=2, stride=2, padding=0)
    total_instructions += x0

    # conv8: (256, 512, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 256, 16, 16) output->(1, 512, 16, 16)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_2_8(class_num=11, in_ch=256, out_ch=512, kernel_size=3, x_shape=(1, 256, 16, 16), PE_array_type0_pair_sparsity=0.897254)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 256*512*3*3

    # conv9: (512, 512, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 512, 16, 16) output->(1, 512, 16, 16)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_2_8(class_num=11, in_ch=512, out_ch=512, kernel_size=3, x_shape=(1, 512, 16, 16), PE_array_type0_pair_sparsity=0.897254)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 512*512*3*3

    # conv10: (512, 512, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 512, 16, 16) output->(1, 512, 16, 16)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_2_8(class_num=11, in_ch=512, out_ch=512, kernel_size=3, x_shape=(1, 512, 16, 16), PE_array_type0_pair_sparsity=0.897254)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 512*512*3*3

    # maxpool4: (kernel_size=2, stride=2) & input->(1, 512, 16, 16) output->(1, 512, 8, 8)
    x0 = estimate_MaxPool2d_layer(x_shape=(1, 512, 16, 16), kernel_size=2, stride=2, padding=0)
    total_instructions += x0

    # conv11: (512, 512, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 512, 8, 8) output->(1, 512, 8, 8)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_2_8(class_num=11, in_ch=512, out_ch=512, kernel_size=3, x_shape=(1, 512, 8, 8), PE_array_type0_pair_sparsity=0.897254)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 512*512*3*3

    # conv12: (512, 512, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 512, 8, 8) output->(1, 512, 8, 8)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_2_8(class_num=11, in_ch=512, out_ch=512, kernel_size=3, x_shape=(1, 512, 8, 8), PE_array_type0_pair_sparsity=0.897254)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 512*512*3*3

    # conv13: (512, 512, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1)) & input->(1, 512, 8, 8) output->(1, 512, 8, 8)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Conv2d_layer_2_8(class_num=11, in_ch=512, out_ch=512, kernel_size=3, x_shape=(1, 512, 8, 8), PE_array_type0_pair_sparsity=0.897254)
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 512*512*3*3

    # maxpool5: (kernel_size=8, stride=1) & input->(1, 512, 8, 8) output->(1, 512, 1, 1)
    x0 = estimate_MaxPool2d_layer(x_shape=(1, 512, 8, 8), kernel_size=8, stride=1, padding=0)
    total_instructions += x0

    # linear1: (512, 11) & input->(1, 512) output->(1, 11)
    x0, x1, x2, x3, x4, x5, x6, x7, x8, x9 = estimate_Linear_layer(class_num=11, in_ch=1, out_ch=11, x_shape=(1, 512))
    total_instructions += x0
    on_chip_quantizer_computation += x1
    data_compressor_computation += x2
    pe_array_type0_computation += x3
    pe_array_type1_computation += x4
    pe_array_type2_computation += x5
    on_chip_dequantizer_computation += x6
    pre_trace_calculator_computation += x7
    post_trace_calculator_computation += x8
    on_chip_mac_computation += x9

    param_size += 512*11 + 11

    return total_instructions, on_chip_quantizer_computation, data_compressor_computation, pe_array_type0_computation, pe_array_type1_computation, pe_array_type2_computation, on_chip_dequantizer_computation, pre_trace_calculator_computation, post_trace_calculator_computation, on_chip_mac_computation, param_size

########################################### Hardware define and evaluation ###########################################
@dataclass
class module:
	area: float   # in um^2
	power: float  # in mW

def print_module_breakdown(title, total_area, total_power, components, area_unit="mm^2"):
	print(f"\n--- {title} Area Breakdown ---")
	for name, comp in components.items():
		area_pct = comp.area / total_area * 100
		if area_unit == "mm^2":
			area_val = comp.area / 1e6
			print(f"{name:<22}: {area_val:.4f} mm^2 ({area_pct:.4f}%)")
		else:
			print(f"{name:<22}: {comp.area:.4f} um^2 ({area_pct:.4f}%)")

	print(f"\n--- {title} Power Breakdown ---")
	for name, comp in components.items():
		power_pct = comp.power / total_power * 100
		print(f"{name:<22}: {comp.power:.4f} mW ({power_pct:.4f}%)")

def print_accelerator_report(name, freq, top_modules, rce_submodules):

	total_area = sum(m.area for m in top_modules.values())
	total_power = sum(m.power for m in top_modules.values())

	print(f"===== {name} Accelerator Report =====")
	print(f"Accelerator frequency: {freq:.2f} GHz")
	print(f"Total area: {total_area/1e6:.4f} mm^2")
	print(f"Total power: {total_power:.4f} mW")

	print("\n--- Component-wise Area Breakdown ---")

	for name_mod, mod in top_modules.items():

		area_mm2 = mod.area / 1e6
		area_pct = mod.area / total_area * 100

		print(f"{name_mod:<24}: {area_mm2:.4f} mm^2 ({area_pct:.4f}%)")

		if name_mod == "RCE":

			for sub_name, sub in rce_submodules.items():
				sub_area = sub.area / 1e6
				sub_pct = sub.area / mod.area * 100

				print(f"  -{sub_name:<21}: {sub_area:.4f} mm^2 ({sub_pct:.4f}%)")

	print("\n--- Component-wise Power Breakdown ---")

	for name_mod, mod in top_modules.items():

		power_pct = mod.power / total_power * 100

		print(f"{name_mod:<24}: {mod.power:.4f} mW ({power_pct:.4f}%)")

		if name_mod == "RCE":

			for sub_name, sub in rce_submodules.items():
				sub_pct = sub.power / mod.power * 100

				print(f"  -{sub_name:<21}: {sub.power:.4f} mW ({sub_pct:.4f}%)")

if __name__ == "__main__":
	clock_frequency = 1.0  # GHz

	# reconfigurable compute engine
	rce = module(area=10961608.5, power=2987.947)
	rce_submodules = {
		"on_chip_quantizer":    module(area=1282371.75, power=473.011),
		"data_compressor":      module(area=319896.0, power=131.041),
		"pe_array_type0":       module(area=399870.0, power=98.14),
		"pe_array_type1":       module(area=79974.0, power=24.162),
		"pe_array_type2":       module(area=389873.25, power=99.218),
		"on_chip_dequantizer":  module(area=1282371.75, power=448.611),
		"pre_trace_calculator": module(area=1831794.75, power=414.767),
		"post_trace_calculator": module(area=2580000.0, power=453.562),
		"fpu":                  module(area=1831794.75, power=414.767),
		"others":               module(area=963662.25, power=430.668),
	}
	top_modules = {
		"RCE":                rce,
		"SIMD Engine":        module(area=2895467.715, power=133.9152),
		"Global SRAM":        module(area=365457.2075, power=235.887),
		"Controller & Others": module(area=1246218.828, power=688.3978),
	}

	print_accelerator_report(
		name="Lonic",
		freq=clock_frequency,
		top_modules=top_modules,
		rce_submodules=rce_submodules
	)

    # other param
	hbm2_energy_per_bit = 5.7 # pJ from nvidia hpca 2017
	timestep = 10
	rce_num = 10
	weight_bitwidth = 4 # int4

	(total_instructions, on_chip_quantizer_computation, data_compressor_computation, pe_array_type0_computation,
	pe_array_type1_computation, pe_array_type2_computation, on_chip_dequantizer_computation, pre_trace_calculator_computation,
	post_trace_calculator_computation, on_chip_mac_computation, param_size) = estimate_vgg_dvs_gesture_training_one_iteration()
    
	#print(param_size)
	#print("total instructions:")
	#print(total_instructions)
	#print("on-chip quantizer:")
	#print(on_chip_quantizer_computation)
	#print("data compressor:")
	#print(data_compressor_computation)
	#print("PE array type0:")
	#print(pe_array_type0_computation)
	#print("PE array type1:")
	#print(pe_array_type1_computation)
	#print("PE array type2:")
	#print(pe_array_type2_computation)
	#print("on-chip dequantizer:")
	#print(on_chip_dequantizer_computation)
	#print("pre-trace calculator:")
	#print(pre_trace_calculator_computation)
	#print("post-trace calculator:")
	#print(post_trace_calculator_computation)
	#print("on-chip mac:")
	#print(on_chip_mac_computation)

    # performance
	print("\n==== Performance ====")

	latency = (total_instructions*timestep+on_chip_quantizer_computation+data_compressor_computation+pe_array_type0_computation+
               pe_array_type1_computation+pe_array_type2_computation+on_chip_dequantizer_computation+pre_trace_calculator_computation+
               post_trace_calculator_computation+on_chip_mac_computation) / (clock_frequency * 1e9) # s
	throughput = 1 / latency # iteration/s

	rce_energy = (rce_submodules["on_chip_quantizer"].power*on_chip_quantizer_computation/(clock_frequency * 1e9)+
                            rce_submodules["data_compressor"].power*data_compressor_computation/(clock_frequency * 1e9)+
                            rce_submodules["pe_array_type0"].power*pe_array_type0_computation/(clock_frequency * 1e9)+
                            rce_submodules["pe_array_type1"].power*pe_array_type1_computation/(clock_frequency * 1e9)+
                            rce_submodules["pe_array_type2"].power*pe_array_type2_computation/(clock_frequency * 1e9)+
                            rce_submodules["on_chip_dequantizer"].power*on_chip_dequantizer_computation/(clock_frequency * 1e9)+
                            rce_submodules["pre_trace_calculator"].power*(pre_trace_calculator_computation+on_chip_mac_computation)/(clock_frequency * 1e9)+
                            rce_submodules["post_trace_calculator"].power*post_trace_calculator_computation/(clock_frequency * 1e9)+
                            rce_submodules["fpu"].power*on_chip_mac_computation/(clock_frequency * 1e9)+
                            rce_submodules["others"].power*latency) * (timestep/rce_num) # mJ
	simd_engine_energy = total_instructions*timestep/(clock_frequency * 1e9)*top_modules["SIMD Engine"].power
	global_sram_energy = latency*top_modules["Global SRAM"].power
	controllers_and_others = latency*top_modules["Controller & Others"].power
	hbm_energy = hbm2_energy_per_bit*(param_size*weight_bitwidth*timestep+param_size*32*timestep+param_size*32*timestep+param_size*32+param_size*weight_bitwidth+param_size*32)*1e-9
	total_energy = (rce_energy+simd_engine_energy+global_sram_energy+controllers_and_others+hbm_energy)
	#print(hbm_energy)
	#print(total_energy)
	print(f"Latency: {latency: .6f} s")
	print(f"Throughput: {throughput: .6f} iteration/s")
	print(f"Energy: {total_energy: .6f} mJ")
```
